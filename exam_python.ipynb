```python
# Q 1

try: 
    lst_nums = []   
    while len(lst_nums) < 10 :
        nums_user = int(input("choose a number:"))
        if nums_user <= 0:
            print("choose only possitive numbers")
            break
        lst_nums.append(nums_user)
    counter_odd = 0
    counter_even = 0    
    for num in lst_nums:
        if num % 2 == 0:
            counter_even += 1
        else:
            counter_odd += 1
    print(f"There are {counter_odd} odd numbers and {counter_even} even numbers")            
except ValueError:
    print("choose only numbers not string") 
```

```python
# Q 2
def permeri_num(n):
    permeri_lst = []
    if n < 0:
        return "choose a positive numbers"
    for i in range(2,n):
        is_prime = True
        for j in range(2,i):
            if i % j == 0:
                is_prime = False
                break
        if is_prime == True:
            permeri_lst.append(i)
    return permeri_lst 
            
print(permeri_num(9))

```

```python
# Q3
try:
    lst_numbers = []
    while True:
        nums = int(input("choose numbers when you want to stop write 0 "))
        if nums < 0:
            print("choose only positive numbers")
            break
        if nums == 0:
            break
        else:
            lst_numbers.append(nums)
    if len(lst_numbers) != 0:
        sum_lst = sum(lst_numbers)
        avg_lst = sum_lst / len(lst_numbers) 
        print(f"The sum of the numbers that the user choose is {sum_lst}, and the average is {avg_lst}") 
    else:
        print("0")           
except ValueError:
    print("choose only numbers")
```

```python
# Q 4
# function open file and write the names of student in each row
def file_studend(file):
      lst_name = []
      while True:
            text = input("enter the name of student(if you wanr stop enter end):")
            if text == "end":
                  break
            else:
                lst_name.append(text)
      str_name = "\n".join(lst_name)                         
      with open(file,"wb") as f:
          f.write(str_name.encode())
      with open(file,"rb") as f:
          return  f.read().decode()    

file = r"C:\Users\salom\OneDrive\שולחן העבודה\cyberpro\python\exam.txt"
print(file_studend(file))     
```
